```python

# Radix Sort using Counting Sort

def counting_sort(arr, exp):
    """
    Counting Sort according to the digit represented by exp.
    exp = 1   -> Ones place
    exp = 10  -> Tens place
    exp = 100 -> Hundreds place
    """

    n = len(arr)

    output = [0] * n
    count = [0] * 10

    # Count occurrences of each digit
    for i in range(n):
        index = (arr[i] // exp) % 10
        count[index] += 1

    # Convert count[] to cumulative count[]
    for i in range(1, 10):
        count[i] += count[i - 1]

    # Build output array (stable sort)
    """
    for i in range(n - 1, -1, -1):
        index = (arr[i] // exp) % 10
        output[count[index] - 1] = arr[i]
        count[index] -= 1
    """
    i = n - 1
    while i >= 0:
        index = arr[i] // exp
        output[count[index % 10] - 1] = arr[i]
        count[index % 10] -= 1
        i -= 1

    # Copy output back into original array
    i=0
    for i in range(0, len(arr)):
        arr[i] = output[i]
def radix_sort(arr):

    max_num = max(arr)

    exp = 1

    while max_num // exp > 0:
        counting_sort(arr, exp)

        if exp == 1:
            print(f"After sorting by Ones place : {arr}")

        elif exp == 10:
            print("After sorting by Tens place :", arr)

        elif exp == 100:
            print("After sorting by Hundreds place:", arr)

        exp *= 10

phone_numbers = [564, 213, 987, 432, 123, 765, 321, 654, 876]

print("Original List")
print(phone_numbers)

print("\nSorting Process")

radix_sort(phone_numbers)

print("\nFinal Sorted List")
print(phone_numbers)
```

## Analysis and Reflection

**How it works.** Radix Sort orders the numbers one digit at a time, starting from the
**least significant digit** (ones), then tens, then hundreds. Each pass uses **Counting Sort**,
which is *stable* - it keeps the order set by the previous, less-significant digit. Once the most
significant digit has been processed, the whole list is sorted.

**Reflection questions**

1. **Order of sorting the digits.** Least Significant Digit (LSD) first - ones, then tens, then
   hundreds, up to the most significant digit of the largest number. Counting Sort's stability is
   what makes sorting in this order produce a correct final result.

2. **After sorting by the least significant digit (ones):**
   `[321, 432, 213, 123, 564, 654, 765, 876, 987]` - the numbers are grouped by their ones digit
   but are not yet fully in order.

3. **After sorting by the most significant digit (hundreds):**
   `[123, 213, 321, 432, 564, 654, 765, 876, 987]` - fully sorted in ascending order.

**Complexity.** `O(d x (n + k))` time - `d` digits, `n` numbers, `k` the digit base (10) - which is
effectively linear for numbers with a fixed number of digits, unlike the `O(n log n)` lower bound of
comparison sorts.

## Order Statistics (background)

Order statistics pick an element by its rank in sorted data: **minimum** (first), **maximum** (last),
**median** (middle), and the **k-th** smallest in general.

- **Quickselect** finds the k-th smallest like QuickSort but recurses into only one side - `O(n)` on
  average, `O(n^2)` worst case.
- **Median of Medians** chooses a good pivot (median of groups of 5) to guarantee `O(n)` selection
  even in the worst case.
- **Uses:** percentiles, top-k selection, median filtering in image processing, and k-NN in machine learning.
